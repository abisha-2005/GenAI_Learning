```python
sentences=["I love Java",
           "I love Python",
           "I like Coding",
           "I like learning",
           "Java is Powerful",
           "Python is Easy"]
print(sentences)
```

**Tokenization**


```python
tokenized_sentence=[]
for sentence in sentences:
  tokens=sentence.split()
  tokenized_sentence.append(tokens)
print(tokenized_sentence)
```

**Vocabulary**

```python
vocabulary=[]
for token in tokenized_sentence:
  for word in token:
    if word not in vocabulary:
      vocabulary.append(word)
print(vocabulary)
```

**Assigning Token ID**

```python
token_words={}
for index,word in enumerate(vocabulary):
  token_words[word]=index
print(token_words)
```

**Sentence to TokenID**

```python
encoded_sentence=[]
for token in tokenized_sentence:
  encode=[]
  for word in token:
    encode.append(token_words[word])
  encoded_sentence.append(encode)
print(encoded_sentence)
```

```python
training_data = []

for sentence in encoded_sentence:
    for i in range(1, len(sentence)):
        input_tokens = sentence[:i]
        target_token = sentence[i]

        training_data.append((input_tokens, target_token))

print(training_data)
```

```python
from collections import defaultdict, Counter

next_word_counts = defaultdict(Counter)

for input_tokens, target_token in training_data:
    context = tuple(input_tokens)
    next_word_counts[context][target_token] += 1

print(dict(next_word_counts))
```

```python
def predict_next(input_text):

    tokens = input_text.split()

    input_ids = []

    for word in tokens:
        input_ids.append(token_words[word])

    context = tuple(input_ids)

    predictions = next_word_counts[context]

    predicted_id = predictions.most_common(1)[0][0]

    for word, word_id in token_words.items():
        if word_id == predicted_id:
            return word


print(predict_next("I love"))
```
